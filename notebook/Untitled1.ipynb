```python
from sentence_transformers import SentenceTransformer

encoder = SentenceTransformer("all-mpnet-base-v2")
vector = encoder.encode(df.text)
```

```python
dim = vector.shape[1]
dim
```

```python
import faiss
index = faiss.IndexFlatL2(dim)
```

```python
import numpy as np
svec = np.array(vec).reshape(1,-1)
svec.shape
```
